exer2: extraindo a tabela do site

```python
import pandas as pd
tab=web.select("table.hockey-stats")[0] #pegando a tabela com a classe "hockey-stats"
tab 
```

```python
col=tab.find("thead").find_all('th') #entra no "thead" e dps pega tudo do 'th'
col_nome=[c.string for c in col]
col_nome #só os nome de cada coluna

tab_lin=tab.find("tbody").find_all('tr') #mesma lógica para pegar as colunas, so que com as tag da linha
l=[]
for tr in tab_lin: #percorre cada elemento da linha, separando e colocando em ordem
    td=tr.find_all('td')
    lin=[str(tr.get_text()).strip() for tr in td] #o strip está limpando os espaçoes em branco 
    l.append(lin)
print(l[0]) #imprime todos os valores da primeira linha
```

```python
df= pd.DataFrame(l, columns=col_nome) #crindo o data frame
df
```

exer3: pegar todas as frases dos "fun facts" que tem a palavra "is"

```python
lis=web.select("ul.fun-facts li")
lis
```

```python
import re
listais=[i.find(string=re.compile('is')) for i in lis] #Percorrendo e vendo qual tem o "is"
listais=[i.find_parent().get_text() for i in listais if i] #aqui ele tira os "none", usa o find_parent para garantir q ta pegando tudo a lista e o get text pra retornar só o texto
listais
```

ex 4: baixar uma imagem da página web

```python
url='https://keithgalli.github.io/web-scraping/' #caminho da url da pág web

imagens=web.select('div.row div.column img') #seleciono as img
img_url=imagens[0]['src']
url_completo=url+img_url #o caminho completo

img_data= requests.get(url_completo).content #tenta acessar o caminho da img
with open("lake_como.jpg",'wb') as handler: #aqui ele abre o caminho usando manipulação de arq em modo de edição
    handler.write(img_data) #write ele "escreve" em bin e acaba criando a img
```

ex5: palavra secreta

```python
files = web.select('div.block a')
relative_files = [f['href'] for f in files] #acessa tudo os links dentro da div secreta

url = "https://keithgalli.github.io/web-scraping/"
for f in relative_files:
    full_url = url + f
    page = requests.get(full_url)
    bs_page = bs(page.content) #ele percorre cada pag html 
    secret_word_element = bs_page.find("p",attrs= {"id": "secret-word"}) #ele percorre pesquisando pelo id os paragrafo de cada uma das pag q achamos
    secret_word = secret_word_element.string #transforma em str pra printar
    print(secret_word)
```
