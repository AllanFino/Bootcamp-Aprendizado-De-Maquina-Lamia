exer6: catálogo de livros (books.toscrape.com)

```python
# site: http://books.toscrape.com/
# 1. pegar todos os "article.product_pod" (cada um é um livro)
# 2. extrair titulo (atributo title da tag <a> dentro do h3), preco (p.price_color) e disponibilidade (p.instock.availability, com get_text().strip())
# 3. montar um dataframe com colunas: titulo, preco, disponibilidade
import pandas as pd
import requests
from bs4 import BeautifulSoup as bs
from collections import Counter
#1
url='http://books.toscrape.com/' #caminho da url da pag
r=requests.get(url)
soup=bs(r.content)
soup

livros=soup.select("article.product_pod") #pega todos os card de livro da pagina
livros
#2
titulo=[l.find("h3").find("a")['title'] for l in livros] #o titulo completo ta no atributo title da tag <a>
titulo

preco=[l.find("p", class_="price_color").get_text().strip() for l in livros] #preco vem com o £ junto
preco
disp=[l.find("p", class_="instock availability").get_text().strip() for l in livros] #tira os espaco em branco
disp
#3
df=pd.DataFrame({"titulo":titulo, "preco":preco, "disponibilidade":disp}) #cria o dataframe
df
```

exer7: baixando algumas capas (books.toscrape.com)

```python
#baixar a capa de todos os livros

url='http://books.toscrape.com/' #caminho da url da pag
r=requests.get(url)
soup=bs(r.content)

livros=soup.select("article.product_pod") #pega todos os card de livro da pag

for i,l in enumerate(livros): #percorre cada livro
    img_url=l.find("img")['src'] #pega o caminho da imagem dentro do card
    url_completo=url+img_url.replace("../","") #o site usa "../" no caminho, tira pra montar certo

    img_data=requests.get(url_completo).content #baixa os dados  da imagem
    with open(f"capa_{i}.jpg",'wb') as handler: #abre um arq novo pro livro
        handler.write(img_data) #'escreve' da imagem no arquivo
```
