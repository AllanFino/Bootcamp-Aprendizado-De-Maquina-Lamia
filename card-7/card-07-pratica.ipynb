```python
import pandas as pd #criando um dataframe simples pra trabalhar um pouco com cada ambiente

df = pd.DataFrame({
    'compra': [10, 20, 15, 30, 25,70],
    'venda':  [15, 25, 18, 40, 33,1]
})

df
```

```python
df['lucro'] = df['venda'] - df['compra'] #criando uma nova coluna
df
```

```python
df['margem_%'] = (df['lucro'] / df['compra']) * 100  # margem percentual
df['status'] = df['lucro'].apply(lambda x: 'lucro' if x > 0 else 'prejuízo') #dando uma brincada com o apply
df
```
