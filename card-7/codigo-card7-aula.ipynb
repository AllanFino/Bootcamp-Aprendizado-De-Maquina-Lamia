```python
print("olá mundo!!")

import pandas as pd
import numpy as np
import matplotlib.pylab as plt
import matplotlib
```

```python
t = np.arange(0.0, 2.0, 0.01)

s = 1 + np.sin(2* np.pi * t)


fig, ax = plt.subplots()
ax.plot(t,s)
#desenhano o gráfico

ax.set(xlabel='time(s)', ylabel = 'voltagem (mV)',
        title = 'titulo super massa')

ax.grid()
```
